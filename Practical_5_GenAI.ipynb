```python
import numpy as np

encoder_outputs = np.array([
    [0.1, 0.2],
    [0.5, 0.3],
    [0.2, 0.8],
    [0.9,  0.1]
])
print(f"\nEncoder Outputs (4 words, 2 features each):\n{encoder_outputs}")


decoder_hidden_state = np.array([0.7, 0.4])
print(f"\nDecoder Hidden State (What we are looking for):\n{decoder_hidden_state}")


alignment_scores = np.dot(encoder_outputs, decoder_hidden_state)
print(f"\nAlignment Scores (Similarity of each encoder to decoder state):{alignment_scores}")

def softmax(x):
  e_x = np.exp(x - np.max(x))
  return e_x/ e_x.sum(axis = 0)

  attention_weights = softmax(alignment_scores)
  print(f"\nAttention Weights (how much to 'Pay attention' to each word): {attention_weights}")
  print(f"(Notice that the word 'student' (index 3) got a higher weight, meaning it's more relevent to 'learning')")


  context_vector = np.sum(encoder_outputs * attention_weights[:, np.newaxis], axis = 0)
  print(f"\nContext Vector (weighted sum of encoder outputs, 'paying attention'):{context_vector}")


  print("\n ---How it Works---")
```
