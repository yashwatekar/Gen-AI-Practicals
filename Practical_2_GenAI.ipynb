```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ['the','king','rules','the','kingdom'],
    ['the','queen','rules','the','kingdom'],
    ['the','prince','rules','the','kingdom'],
    ['an','apple','is','a','fruit'],
    ['a','banana','is','a','fruit'],
    ['an','orange','is','a','fruit'],
]

model_cbow = Word2Vec(sentences, min_count=1, vector_size=100, window=2, sg=0, epochs=100)
print("CBOW model trained successfully")

model_skipgram = Word2Vec(sentences, min_count=1, vector_size=100, window=2, sg=1, epochs=100)
print("Skip-gram model trained successfully")

def test_similarites(model, model_name):
  print(f"--- Semantic Similarities using {model_name} ---")

  sim_royal = model.wv.similarity('queen', 'king')
  print(f"Similarities ('queen','king'): {sim_royal:.4f}")

  sim_fruit = model.wv.similarity('apple', 'is')
  print(f"Similarities ('apple','is'): {sim_fruit:.4f}")

  sim_unrelated = model.wv.similarity('queen', 'apple')
  print(f"Similarity of unrelated words ('queen','apple'): {sim_unrelated:.4f}")
  print("-" * 40)

test_similarites(model_cbow, "CBOW")
test_similarites(model_skipgram, "Skip-gram")
```
